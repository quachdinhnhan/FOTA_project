```python
import sys
from pathlib import Path

import yaml
import librosa
import librosa.display
import matplotlib.pyplot as plt
import numpy as np

PROJECT_ROOT = Path('/Users/apple/Downloads/HUTECH/Thesis/FOTA_project').resolve()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from src.attacks.jingleback import JingleBackAttack

with open(PROJECT_ROOT / 'configs' / 'attacks' / 'jingleback.yaml', 'r') as f:
    jingleback_cfg = yaml.safe_load(f)

SR = jingleback_cfg['sr']
HOP_LENGTH = 256
WIN_LENGTH = 1024
N_FFT = 1024
N_MELS = 80

TRAIN_LEFT_DIR = PROJECT_ROOT / 'datasets' / 'raw' / 'train' / 'left'
raw_wav_files = sorted(TRAIN_LEFT_DIR.glob('*.wav'))
if not raw_wav_files:
    raise FileNotFoundError(f'No WAV files found in {TRAIN_LEFT_DIR}')

sample_wav = raw_wav_files[0]
original_audio, _ = librosa.load(sample_wav, sr=SR)

jingleback = JingleBackAttack(
    target_label=0,
    poison_rate=jingleback_cfg['poison_rate'],
    sr=SR,
    pitch_semitones=jingleback_cfg['pitch_semitones'],
    distortion_db=jingleback_cfg['distortion_db'],
    chorus_rate_hz=jingleback_cfg['chorus_rate_hz'],
    chorus_depth=jingleback_cfg['chorus_depth'],
    chorus_center_delay_ms=jingleback_cfg['chorus_center_delay_ms'],
    chorus_feedback=jingleback_cfg['chorus_feedback'],
    chorus_mix=jingleback_cfg['chorus_mix'],
)
triggered_audio = jingleback.inject_trigger(original_audio)

original_mel = librosa.feature.melspectrogram(
    y=original_audio,
    sr=SR,
    n_fft=N_FFT,
    hop_length=HOP_LENGTH,
    win_length=WIN_LENGTH,
    n_mels=N_MELS,
)
triggered_mel = librosa.feature.melspectrogram(
    y=triggered_audio,
    sr=SR,
    n_fft=N_FFT,
    hop_length=HOP_LENGTH,
    win_length=WIN_LENGTH,
    n_mels=N_MELS,
)

original_db = librosa.power_to_db(original_mel, ref=np.max)
triggered_db = librosa.power_to_db(triggered_mel, ref=np.max)

print(f'Using sample: {sample_wav.name}')
print(f'Original shape: {original_db.shape}')
print(f'JingleBack shape: {triggered_db.shape}')
print(f'Original dB range: [{original_db.min():.2f}, {original_db.max():.2f}]')
print(f'JingleBack dB range: [{triggered_db.min():.2f}, {triggered_db.max():.2f}]')
```

```python
plt.rcParams.update({
    'font.family': 'serif',
    'font.size': 10,
    'axes.labelsize': 10,
    'xtick.labelsize': 10,
    'ytick.labelsize': 10,
})

fig, ax = plt.subplots(figsize=(5, 4))
librosa.display.specshow(
    triggered_db,
    sr=SR,
    hop_length=HOP_LENGTH,
    x_axis='time',
    y_axis='mel',
    fmax=SR / 2,
    cmap='viridis',
    ax=ax,
)
ax.set_xlabel('Time (s)')
ax.set_ylabel('Frequency (Hz)')
# ax.set_title('JingleBack Triggered Sample')
ax.set_xticks([0, 0.3, 0.6, 0.9])
ax.set_yticks([0, 512, 1024, 2048, 4096, 8192])
plt.tight_layout()
plt.show()
```
