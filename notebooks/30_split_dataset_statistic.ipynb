```python
from pathlib import Path

base_dir = Path("../datasets/raw")

for split in ["train", "test", "val"]:
    split_dir = base_dir / split
    file_count = sum(1 for p in split_dir.rglob("*") if p.is_file())
    print(f"{split}: {file_count}")
    # print total number of files in each class subdirectory
    for class_dir in sorted(split_dir.iterdir()):
        if class_dir.is_dir():
            class_file_count = sum(1 for p in class_dir.rglob("*") if p.is_file())
            print(f"  {class_dir.name}: {class_file_count}")
```
