```python
from pathlib import Path

base_dir = Path("../datasets/npy_scd10")  # Change this to your dataset directory

for split in ["train", "val", "test"]:
    split_dir = base_dir / split
    if not split_dir.exists():
        print(f"Split directory not found: {split_dir.resolve()}")
        continue

    class_dirs = sorted(p for p in split_dir.iterdir() if p.is_dir())
    class_counts = {class_dir.name: len(list(class_dir.glob("*.npy"))) for class_dir in class_dirs}

    print(f"\nSplit: {split}")
    print(f"Number of classes: {len(class_dirs)}")
    print(f"Total files: {sum(class_counts.values())}")
    print("Per-class file counts:")
    for class_name, count in class_counts.items():
        print(f"{class_name}: {count}")
```
