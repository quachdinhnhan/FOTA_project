```python
from pathlib import Path

# Dataset root (relative to the notebooks/ folder)
DATASET_ROOT = Path("../datasets/raw/speech_commands")

if not DATASET_ROOT.exists():
    raise FileNotFoundError(f"Dataset directory not found: {DATASET_ROOT.resolve()}")

class_dirs = sorted(p for p in DATASET_ROOT.iterdir() if p.is_dir())

print(f"Dataset root: {DATASET_ROOT.resolve()}")
```

```python
# Calculate the number of files in each class directory
class_counts = {class_dir.name: len(list(class_dir.glob("*.wav"))) for class_dir in class_dirs}

print(f"Number of classes: {len(class_dirs)}")
print(f"Total files: {sum(class_counts.values())}")
print("\nPer-class file counts:")
for class_name, count in class_counts.items():
    print(f"{class_name}: {count}") 

```
